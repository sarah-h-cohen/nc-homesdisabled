# Re-analysis of Buzzfeed's 2022 ICF story


Buzzfeed focused on the intermediate care facilities owned by KKR, the hedge fund that for decades has been known to take over struggling industries. We're more interested in a general analysis of the quality of North Carolina homes, how they compare nationally, and whather those owned by RHS are different than other for-profit, non-profit or government. 

Buzzfeed's data includes the following: 

* Ownership data for six states, including North Carolina. It's very focused on getting KKR right, not on everyone. But for North Carolina, it's pretty much the same as the data that is provided in download. 

* QCOR survey inspection data set, which they obtained from CMS. I have to look to see if it's national or if it just pulled the states we wanted. 

The data only goes through 2021, but we can use it to get a first cut of the information they have, and see whether our story would have been useful at the time.
